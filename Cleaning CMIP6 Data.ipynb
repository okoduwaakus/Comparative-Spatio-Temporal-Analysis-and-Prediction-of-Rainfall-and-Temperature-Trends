```
import pandas as pd
```

```
# 1. Load CSV
df = pd.read_csv("/content/drive/MyDrive/Raw CMIP6/CMIP6_MPI-ESM1-2-LR_ssp585.csv")
```

```
df.head(2)
```

```
# 2. Create a proper date column
df["date"] = pd.to_datetime(dict(year=df.year, month=df.month, day=1))
df = df.sort_values("date")
```

```
# 3. Replace zeros rainfall values
# That is, the zeros is what i now call missing
df["pr_mm_day"] = df["pr_mm_day"].replace(0, pd.NA)
```

```
# 4. Interpolate missing rainfall
df["pr_mm_day"] = df["pr_mm_day"].interpolate(method="linear")
```

```
# 5. Fill remaining edges
df["pr_mm_day"] = df["pr_mm_day"].fillna(method="bfill").fillna(method="ffill")
```

```
# 6. Save cleaned file
df.to_csv("CMIP6_MPI-ESM1-2-LR_ssp585_cleaned.csv", index=False)
```
